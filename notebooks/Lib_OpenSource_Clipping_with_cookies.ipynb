### 🍪 (Opsional) Upload cookies.txt
Diperlukan hanya untuk video YouTube yang **age-restricted** atau **login-required**. Untuk video publik biasa, cell ini boleh diskip (klik ⏹ Stop saat dialog upload muncul, atau langsung lanjut ke cell berikutnya tanpa menjalankan cell ini).

```python
# Upload cookies.txt (opsional)
from google.colab import files
import os

COOKIES_FILE = None

print("📤 Silakan pilih file cookies.txt (format Netscape, bisa didapat via ekstensi 'Get cookies.txt LOCALLY').")
try:
    uploaded = files.upload()
    if uploaded:
        cookies_filename = list(uploaded.keys())[0]
        if cookies_filename != "cookies.txt":
            os.rename(cookies_filename, "cookies.txt")
        COOKIES_FILE = os.path.abspath("cookies.txt")
        print(f"✅ cookies.txt berhasil diupload: {COOKIES_FILE}")
except Exception as e:
    print(f"⚠️ Upload dilewati: {e}")

if not COOKIES_FILE:
    print("ℹ️ Tidak ada cookies.txt — download akan berjalan tanpa autentikasi (normal untuk video publik).")
```
